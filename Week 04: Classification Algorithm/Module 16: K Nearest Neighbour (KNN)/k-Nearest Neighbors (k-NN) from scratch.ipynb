# 1. Calculating Distance

```python
import math
def euclidean_distance(point1, point2):
    distance = 0.0
    for i in range(len(point1)):
        distance += (point1[i] - point2[i]) ** 2
    return math.sqrt(distance)
```

# 2 & 3. Sorting & Selecting the Neighbors

```python
def get_neighbors(train_x,train_y,qury_point,k):
    distances=[]
    for i in range(len(train_x)):
        dist=euclidean_distance(qury_point,train_x[i])
        distances.append((train_x[i],train_y[i],dist))
    distances.sort(key=lambda x: x[2])
    neighbors=[]
    for i in range(k):
        neighbors.append(distances[i])
    return neighbors
```

# 4. Classifying with the Dominant Class

```python
from collections import Counter
def predict_classification(train_x,train_y,qury_point,k):
    neighbors=get_neighbors(train_x,train_y,qury_point,k)
    output_values=[row[1] for row in neighbors]
    prediction=Counter(output_values).most_common(1)
    return prediction[0][0]
```

# Testing

```python
# Toy Dataset: [Weight, Roundness]
X_train = [
    [150, 8],  # Apple
    [160, 9],  # Apple
    [170, 3],  # Orange
    [180, 2],  # Orange
]
y_train = ["Apple", "Apple", "Orange", "Orange"]

unknown_fruit = [155, 7]
k = 3

# Run prediction
prediction = predict_classification(X_train, y_train, unknown_fruit, k)
print(f"The predicted class for the unknown fruit is: {prediction}")
```
