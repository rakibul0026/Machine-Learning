```python
df=pd.read_csv('Iris.csv')
df
```

```python
x=df.drop('Species',axis=1)
y=df['Species']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
knn_model=Pipeline(
    steps=[
        ('scaler',StandardScaler()),
        ('knn',KNeighborsClassifier())
        
    ]
)
knn_model.fit(x_train,y_train)
```

```python
predict=knn_model.predict(x_test)
predict
```

```python
accuracy_score=accuracy_score(y_test,predict)
print('Accuracy:',accuracy_score)
```

```python
confusion_matrix=confusion_matrix(y_test,predict)
print('Confusion Matrix:\n',confusion_matrix)
```

```python
report=classification_report(y_test,predict)
print('Classification Report:\n',report)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.datasets import load_wine

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score

from sklearn.metrics import confusion_matrix,classification_report

from sklearn.model_selection import GridSearchCV
```

```python
df=pd.read_csv('marketing_campaign.csv', sep='\t')
df.head(5)
```

```python
df.isnull().sum()
```

```python
df['Income'].fillna(df['Income'].mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
x=df.select_dtypes(include=[np.number]).drop('Response',axis=1)
y=df['Response']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
pipeline=Pipeline(
    steps=[
        ('scaler',StandardScaler()),
        ('knn',KNeighborsClassifier())
        
    ]
)
pipeline.fit(x_train,y_train)
```

```python
grid_params={
    'knn__n_neighbors': [3, 5, 7, 9],
    'knn__weights': ['uniform', 'distance'],
    'knn__p': [1, 2]
    
}
grid_search=GridSearchCV(pipeline,grid_params,cv=5)
grid_search.fit(x_train,y_train)
```

```python
best_params=grid_search.best_params_
print('Best Parameters:',best_params)
```

```python
#Compare tuned KNN with default KNN.
knn_predict=grid_search.predict(x_test)
knn_accuracy=accuracy_score(y_test,knn_predict) 

default_knn = KNeighborsClassifier()
default_knn.fit(x_train, y_train)
predict_default = default_knn.predict(x_test)
default_accuracy = accuracy_score(y_test, predict_default)
print("Tuned KNN Accuracy:", knn_accuracy)
print("Default KNN Accuracy:", default_accuracy)
```

```python
precision = precision_score(y_test, knn_predict)
recall = recall_score(y_test, knn_predict)
print("Precision:", precision)
print("Recall:", recall)
```

```python
#F1 score
from sklearn.metrics import f1_score
f1 = f1_score(y_test, knn_predict)
print("F1 Score:", f1)
```
