```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
```

```python
x=np.array([
    [4,5,45],
    [3,5,60],
    [4,5,60],
    [3,2,56],
    [7,8,90],
    [2,5,45],
    [3,5,60],
    [1,5,60],
    [3.5,2,56],
    [8,8,90]
    
])
y_actual=np.array([2000,34000,3500,4500,7600,2300,3400,5500])
def make_predict(x,w,b):
    m=x.shape[0]
    predict_list=np.zeros((m,))
    for i in range(m):
        predict_list[i]=np.dot(w,x[i])+b
        
    return predict_list
```

```python
m=x.shape[0]
n=x.shape[1]

w_init=np.ones((n,))
b_init=1.0

y_pred=make_predict(x,w_init,b_init)

print(f"{'Actual (আসল)':<15} | {'Predicted (অনুমান)':<15}")
print("-" * 40)
for act, pred in zip(y_actual, y_pred):
    print(f"{act:<15.2f} | {pred:<15.2f}")
```
