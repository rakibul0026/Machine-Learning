```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split


from sklearn.datasets import load_breast_cancer

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier


from sklearn.metrics import accuracy_score,precision_score,recall_score

from sklearn.metrics import confusion_matrix,classification_report

from sklearn.model_selection import GridSearchCV
```

# 1. Decision Tree: Overfitting vs Underfitting

```python
df=pd.read_csv('Iris.csv')
df
```

```python
x=df.drop(['Species'],axis=1)
y=df['Species']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=DecisionTreeClassifier(max_depth=2)
model.fit(x_train,y_train)

y_pred=model.predict(x_train)
print('Train Accuracy:',accuracy_score(y_train,y_pred))

y_pred_test=model.predict(x_test)
print('Test Accuracy:',accuracy_score(y_test,y_pred_test))

```

```python
model=DecisionTreeClassifier()
model.fit(x_train,y_train)

y_pred=model.predict(x_train)
print('Train Accuracy:',accuracy_score(y_train,y_pred))

y_pred_test=model.predict(x_test)
print('Test Accuracy:',accuracy_score(y_test,y_pred_test))

```

Which model is underfitting/overfitting? 

Model A (max_depth=2) risks underfitting if it doesn't capture enough complexity (though Iris is simple enough that it still does well). Model B (max_depth=None) achieves 100% training accuracy but risks overfitting on noisier datasets because it builds deep rules to perfectly isolate every training sample.
