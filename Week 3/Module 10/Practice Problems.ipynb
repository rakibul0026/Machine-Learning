Problem 1: House Price Prediction using Gradient Descent 

Dataset: 

House Prices Dataset (Kaggle) 

Tasks 

1.  Load the dataset using Pandas. 

2.  Select one feature (e.g., area) and the target variable (price). And convert them to numpy array 

3.  Implement the Mean Squared Error (MSE) Cost Function manually. 

4.  Calculate the initial cost using random values of slope and intercept. 

5.  Apply Gradient Descent from scratch.

6.  Train for 1000 iterations. 

7.  Plot: 

        ○  Iteration vs Cost 

        ○  Actual vs Predicted Values 
   
8.  Report the final learned parameters

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

#Load the dataset using Pandas. 
df=pd.read_csv('Housing.csv')
df
```

```python
# Select one feature (e.g., area) and the target variable (price). And convert them to numpy array 
area = df['area'].values
print("Area:\n", area)
price=df['price'].values
print("Price:\n", price)
```

```python
#Implement the Mean Squared Error (MSE) Cost Function manually
def compute_cost(x, y, w, b):
    m = len(x)
    total_cost = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost = (f_wb - y[i]) ** 2
        total_cost += cost
    return total_cost / (2 * m)
```

```python
#Calculate the initial cost using random values of slope and intercept. 
initial_w=0.0
initial_b=0.0
initial_cost=compute_cost(area,price,initial_w,initial_b)
print("Initial Cost:", initial_cost)
```

```python
#Apply Gradient Descent from scratch.
def compute_gradient(x, y, w, b):
    m = len(x)
    dj_dw = 0
    dj_db = 0
    for i in range(m):
        f_wb = w * x[i] + b
        error = f_wb - y[i]
        dj_dw += error * x[i]
        dj_db += error
    dj_dw /= m
    dj_db /= m
    return dj_dw, dj_db
```

```python
#Train for 1000 iterations. 
def gradient_descent(x, y, w_in, b_in, alpha, num_iters):
    costs=[]
    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(x, y, w_in, b_in)
        w_in -= alpha * dj_dw
        b_in -= alpha * dj_db
        cost = compute_cost(x, y, w_in, b_in)
        costs.append(cost)
        if i % 100 == 0:
            print(f"Iteration {i}: Cost {cost}, w {w_in}, b {b_in}")
    return w_in, b_in, costs
```

```python
w=initial_w
b=initial_b
alpha = 0.0001
num_iters = 1000
w_final, b_final, costs = gradient_descent(area, price, w, b, alpha, num_iters)
print("Final parameters: w =", w_final, ", b =", b_final)
```

```python
#plot cost
plt.plot(costs)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Iteration vs Cost")
plt.show()
```

```python
#actual vs predicted
y_pred = w_final * area + b_final
plt.scatter(area, price, color='blue', label='Actual')
plt.scatter(area, y_pred, color='red', label='Predicted')
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Actual vs Predicted")
plt.legend()
plt.show()
```

5.  Print the cost after every 50 iterations. 

6.  Plot Cost vs Iteration.

```python
df=pd.read_csv("StudentsPerformance.csv")
x=df['reading score'].values
y=df['writing score'].values

def compute_cost(x,y,w,b):
    m=len(x)
    total_cost=0
    for i in range(m):
        f_wb=w*x[i]+b
        cost=(f_wb-y[i])**2
        total_cost+=cost
    return total_cost/(2*m)
```

```python
def compute_gradient(x,y,w,b):
    m=len(x)
    dj_dw=0
    dj_db=0
    for i in range(m):
        f_wb=w*x[i]+b
        error=f_wb-y[i]
        dj_dw+=error*x[i]
        dj_db+=error
    dj_dw/=m
    dj_db/=m
    return dj_dw,dj_db
    
```

```python
w=0.0
b=0.0
alpha=0.0001
num_iters=500
costs=[]
for i in range(num_iters):
    dj_dw,dj_db=compute_gradient(x,y,w,b)
    w-=alpha*dj_dw
    b-=alpha*dj_db
    cost=compute_cost(x,y,w,b)
    costs.append(cost)
    if i%50==0:
        print(f"Iteration {i}: Cost {cost}, w {w}, b {b}")
```

```python
plt.plot(costs)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Iteration vs Cost")
```

Problem 4: Salary Prediction \

Dataset: 

Salary Dataset 

Tasks 

1.  Predict Salary from YearsExperience. 

2.  Initialize parameters randomly.

3.  Calculate cost before training. 

4.  Apply Gradient Descent.

5.  Plot: 

   ○  Cost vs Iteration 
   
   ○  Regression Line

```python
df=pd.read_csv("Salary_Data.csv")
df
```

```python
x=df["YearsExperience"].values
y=df["Salary"].values

def cost(x,y,w,b):
    m=len(x)
    return np.sum((w*x+b-y)**2)/(2*m)
```

```python
def compute_gradient(x,y,w,b):
    m=len(x)
    y_pred=w*x+b
    
    dw=np.sum((y_pred-y)*x)/m
    db=np.sum(y_pred-y)/m
    return dw,db
```

```python
# Random Initialization
w=np.random.randn()
b=np.random.randn()
alpha=0.01
num_iters=1000
costs=[]
for i in range(num_iters):
    dw,db=compute_gradient(x,y,w,b)
    w-=alpha*dw
    b-=alpha*db
    costs.append(cost(x,y,w,b))
print("Final w =",w)
print("Final b =",b)
```

```python
plt.plot(costs)
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration")
plt.show()
```

```python
y_pred=w*x+b
plt.scatter(x,y,color="blue",label="Actual")
plt.scatter(x,y_pred,color="red",label="Predicted")
plt.title("Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

Problem 5: Car Fuel Consumption Prediction 

Dataset: 

Auto MPG Dataset 

Tasks 

1.  Use horsepower to predict mpg. 

2.  Implement MSE Cost Function manually. 

3.  Apply Gradient Descent. 

4.  Train for different learning rates: 

       ○  0.001 

       ○  0.01 

       ○  0.1 
       
5.  Compare the convergence for different learning rates.

```python
df=pd.read_csv("auto-mpg.csv")
df
```

```python
x=df["horsepower"].values
y=df["mpg"].values

def cost(x,y,w,b):
    m=len(x)
    return np.sum((w*x+b-y)**2)/(2*m)
def gradient(x,y,w,b):
    m=len(x)

    y_pred=w*x+b

    dw=np.sum((y_pred-y)*x)/m
    db=np.sum(y_pred-y)/m

    return dw,db
```

```python
learning_rates=[0.001,0.01,0.1]

for alpha in learning_rates:

    w=0
    b=0

    costs=[]

    for i in range(1000):

        dw,db=gradient(x,y,w,b)

        w-=alpha*dw
        b-=alpha*db

        costs.append(cost(x,y,w,b))

    plt.plot(costs,label=f"LR={alpha}")

plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Learning Rate Comparison")
plt.legend()
plt.show()
```
