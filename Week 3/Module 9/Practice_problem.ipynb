### Task 5

```python
# sklearn ব্যবহার করে একটি simple linear regression model তৈরি করো
from sklearn.linear_model import LinearRegression
import numpy as np
# Sample data
x=np.array([1000,1500,2000,3500,4000]).reshape(-1,1)  # Reshape for sklearn
y=np.array([300000,400000,500000,650000,700000])

# Create and train the model
model = LinearRegression()
model.fit(x, y)

# Make a prediction
house_size = 2500
predicted_price = model.predict([[house_size]])
print(predicted_price)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

x=np.array([[1000],[1500],[2000],[3500],[4000]])
y=np.array([300000,400000,500000,650000,700000])

model=LinearRegression()
model.fit(x, y)

house_size=2500
predicted_price=model.predict([[house_size]])
print(predicted_price)
```

"What should an employee's salary be based on his years of experience?"

```python
x_experience=np.array([1,2,3,4,5,6,7]).reshape(-1,1)
y_salary=np.array([30000,40000,50000,65000,70000,80000,90000])

model=LinearRegression()
model.fit(x_experience, y_salary)

#if  experience is 8 years, what will be the predicted salary?
experience=8
predicted_salary=model.predict([[experience]])


print(predicted_salary)
```

```python
plt.scatter(x_experience, y_salary, color='blue', label='Data Points')
plt.plot(x_experience, model.predict(x_experience), color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.legend()
plt.show()
```
