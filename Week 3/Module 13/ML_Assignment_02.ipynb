## Question 2 (10 Marks)

Handle missing values and perform feature engineering:
- Impute missing numerical values using `SimpleImputer` with mean strategy
- Impute missing categorical values using most frequent strategy
- Drop columns with more than 50% missing values
- Perform train-test split with `test_size=0.2` and `random_state=42`

Display the shape of final train and test sets.

```python
#Question 2

from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

#Drop columns with more than 50% missing values
missing_percent=df.isnull().sum()/len(df)*100
cols_to_drop=missing_percent[missing_percent>50].index
df_clean=df.drop(columns=cols_to_drop)

x=df_clean.drop(columns=['House_Price'])
y=df_clean['House_Price']

num_cols=['Square_Footage','Num_Bedrooms','Num_Bathrooms','Year_Built','Lot_Size','Garage_Size']
cat_cols=['Neighborhood_Quality']

num_imputer=SimpleImputer(strategy="mean")
x[num_cols]=num_imputer.fit_transform(x[num_cols])

if len(cat_cols)>0:
    cat_imputer=SimpleImputer(strategy="most_frequent")
    x[cat_cols]=cat_imputer.fit_transform(x[cat_cols])
    
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

print("X_train shape:",x_train.shape)
print("X_test shape:",x_test.shape)
print("y_train shape:",y_train.shape)
print("y_test shape:",y_test.shape)

```

## Question 3 (20 Marks)

Implement **Simple Linear Regression** using **only NumPy** (no Scikit-Learn allowed):
- Compute slope (`m`) and intercept (`c`) using the Batch Gradient Descent
- Predict values for the test set
- Print the learned `m` and `c` values

Use `Square_Footage` as feature (X) and `House_Price` as target (y).

```python
#Question 3
import numpy as np
x=x_train['Square_Footage'].values
y = y_train.values

x_test_simple=x_test['Square_Footage'].values

x_mean,x_std=np.mean(x),np.std(x)
x_scaled=(x-x_mean)/x_std
x_test_scaled=(x_test_simple-x_mean)/x_std
def compute_cost(x,y,m,c):
    n=len(y)
    y_pred=m*x+c
    cost=(1/n)*np.sum((y_pred-y)**2)
    return cost

def gradient(x,y,m,c):
    n=len(y)
    y_pred=m*x+c
    dw=(2/n)*np.sum((y_pred-y)*x)
    dc=(2/n)*np.sum(y_pred-y)
    return dw, dc
def  gradient_descent(x, y, m_in, c_in, alpha, num_iters):
    m=m_in
    c=c_in
    for i in range(num_iters):
        dw,dc=gradient(x,y,m,c)
        m=m-alpha*dw
        c=c-alpha*dc
        if i % 100 == 0:
         cost=compute_cost(x,y,m,c)
         print(f"Iteration {i},cost:{cost}")
    return m,c
m_init=0.0
c_init=0.0
m_scaled,c_scaled=gradient_descent(x_scaled,y,m_init,c_init,alpha=0.01,num_iters=1000)
m=m_scaled/x_std
c=c_scaled-(m_scaled*x_mean/x_std)

y_pred=m*x_test_simple+c
print("predicted house prices:",y_pred[:5])

print("m:",m)
print("c:",c)
```

## Question 4 (10 Marks)

Build a **ColumnTransformer** that applies:
- `StandardScaler` on numerical columns: `Square_Footage`, `Num_Bedrooms`, `Num_Bathrooms`
- `OneHotEncoder` on categorical column: `Neighborhood_Quality`



```python
#Question 4
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

num_cols=['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms']
cat_cols=['Neighborhood_Quality']

preprocessor=ColumnTransformer(
    transformers=[
      ('num',StandardScaler(), num_cols),
      ('cat',OneHotEncoder(),cat_cols)
      ]
)
print(preprocessor)
```

## Question 5 (20 Marks)

Build a complete **Pipeline** using Scikit-Learn that includes:
- The `ColumnTransformer`
- `SGDRegressor` as the final estimator
- Train the pipeline and evaluate using RMSE and R² score
- Print predicted vs actual values for the first 10 test samples

```python
# Question 5
from sklearn.pipeline import Pipeline
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error, r2_score
pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('regressor',SGDRegressor(max_iter=1000,random_state=42))
    
])
pipeline.fit(x_train,y_train)
y_pred=pipeline.predict(x_test)
rmse=np.sqrt(mean_squared_error(y_test,y_pred))
r2=r2_score(y_test,y_pred)
print(f"RMSE:{rmse}")
print(f"R2 Score:{r2}")

print("First 10 predicted house prices:",y_pred[:10])
print("First 10 actual house prices:",y_test[:10])

```

## Question 7 (10 Marks) (You have to explore the topic and use the equation via Numpy)

Implement **Multiple Linear Regression** using **only NumPy**:
- Pick random 100 datas from the dataset
- Use the Normal Equation: `θ = (XᵀX)⁻¹ Xᵀy`
- Use `Square_Footage`, `Num_Bedrooms`, and `Num_Bathrooms` as features
- Print the learned coefficients (θ values)

```python
# Question 7
import numpy as np
df_sample=df.sample(n=100,random_state=42)
print(df_sample)
```

```python
features=['Square_Footage' , 'Num_Bedrooms' ,'Num_Bathrooms']
x_np=df_sample[features].values
y_np=df_sample['House_Price'].values

x_bias=np.c_[np.ones(x_np.shape[0]),x_np]

theta=np.linalg.inv(x_bias.T .dot(x_bias)).dot(x_bias.T).dot(y_np)

print("Learned Coefficients(Theta values):",theta)
print("Intercept:",theta[0])
print("Square Footage Coefficient:",theta[1])
print("Num Bedrooms Coefficient:",theta[2])
print("Num Bathrooms Coefficient:",theta[3])
```
