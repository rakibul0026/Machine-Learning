outlier Detection

using Z-score

```python
mean_of_age=X_train['Age'].mean()
std_of_age=X_train['Age'].std()

X_train['Z_score_age']=(X_train['Age']-mean_of_age)/std_of_age
X_test['Z_score_age']=(X_test['Age']-mean_of_age)/std_of_age
X_train
```

```python
outlier_age=X_train[abs(X_train['Z_score_age'])>3]
print(len(outlier_age))
display(outlier_age)
```

```python
mean_of_fare=X_train['Fare'].mean()
std_of_fare=X_train['Fare'].std()

X_train['Z_score_fare']=(X_train['Fare']-mean_of_fare)/std_of_fare
#X_test['Z_score_fare']=(X_test['Fare']-mean_of_fare)/std_of_age
X_train
```

```python
outlier_fare=X_train[abs(X_train['Z_score_fare'])>3]
print(len(outlier_fare))
display(outlier_fare)
```

Outliers with IQR

```python
age_Q1=X_train['Age'].quantile(0.25)
age_Q3=X_train['Age'].quantile(0.75)

age_IQR=age_Q3-age_Q1

age_minimum=age_Q1-1.5*age_IQR
age_maximum=age_Q3+1.5*age_IQR

print(age_minimum,age_maximum)
```

```python
age_outlier_IQR=X_train[(X_train['Age']<age_maximum)]
print(len(age_outlier_IQR))
display(age_outlier_IQR)
```

```python
#fare
fare_Q1=X_train['Fare'].quantile(0.25)
fare_Q3=X_train['Fare'].quantile(0.75)

fare_IQR=fare_Q3-fare_Q1
fare_minimum=max(fare_Q1-1.5*fare_IQR, X_train['Fare'].min())
fare_maximum=fare_Q3+1.5*fare_IQR
print(fare_minimum,fare_maximum)
```

```python
fare_outlier_IQR=X_train[(X_train['Fare']<fare_maximum)]
print(len(fare_outlier_IQR))
display(fare_outlier_IQR)
```

```python
X_train=X_train[abs(X_train['Z_score_fare']<=3)]
X_train
```

```python
minimum_zscore_fare=X_train[abs(X_train['Z_score_fare']<=3)]['Fare'].max()
print(minimum_zscore_fare)
```

```python
X_train['Fare']=X_train['Fare'].clip(fare_minimum,fare_maximum)
X_train['Fare'].max()
```
